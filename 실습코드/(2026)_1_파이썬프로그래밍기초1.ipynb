## 1.1 숫자형
- 숫자형(Number)이란 숫자 형태로 이루어진 자료형으로, 이미 잘 알고 있는 것들 입니다. 숫자형으로 사용되는 항목에는 정수, 실수, 8진수, 16진수, 복소수 형태가 있습니다.
- 그 중, 정수형과 실수형의 형태가 가장 많이 활용되는 형태라고 할 수 있습니다.

### 1.1.1. 정수형

```python
print(123)
print(456)
```

```python
print(123) # 정수 출력
```

### 1.1.2. 실수형

```python
print(123.45) # 실수 출력
```

### 1.1.5. 숫자형을 활용하기 위한 연산자
- 숫자형(Number)을 활용하기 위해 기본적인 사칙 연산 뿐만 아니라 다양한 형태의 연산자를 사용할 수 있습니다.


```python
print(10+5) # 덧셈
print(10-5) # 뺄셈
print(10*5) # 곱셈
print(10/5) # 나눗셈

print(234562*1234526)
```

```python
print(15/2)  # 나눗셈
print(15//2) # 몫
print(15%2)  # 나머지
```

```python
# math 라이브러리(공구함)를 불러옵니다.
import math

# math.sqrt() : 제곱근을 구해줍니다.
print("16의 제곱근:", math.sqrt(16))
# math.pi : 원주율(π) 값을 알려줍니다.
print("원주율:", math.pi)
# math.ceil() : 숫자를 정수로 '올림'해줍니다.
print("2.1의 올림:", math.ceil(2.1))
# math.floor() : 숫자를 정수로 '내림'해줘요.
print("2.9의 내림:", math.floor(2.9))
```

### 1.2.2. 문자열 연산

문자열은 방마다 번호(인덱스)가 매겨져 있기 때문에, 특정 방에 있는 글자 하나만 콕 집어내거나, 몇 번 방부터 몇 번 방까지의 글자들을 잘라낼 수 있습니다.

```python
print("안녕하세요"[:2])
```

```python
# 문자열 더해서 연결하기 (Concatanation)
# + 연산자
print("안녕" + "하세요")

head = "Python"
tail = " is fun!"
print(head + tail)

head2 = "Python"
tail2 = "is fun!"
print(" ".join([head2, tail2]))
```

```python
# - 연산자
print("Hello" - "H")
```

```python
# 문자열 곱하기
# * 연산자
print("안녕" * "하세요")
print("화이팅!" * 3)

a = "python"
a*2

b = "python "
b*2

#(b*2).strip()
```

```python
# / 연산자
print("Hello" / "H")
print("파이썬 화이팅!" / 3)
```

```python
# 문자열 인덱싱
print("Python"[0])
print("Python"[3])
```

```python
"Life is too short, You need python"[-7:-3]
```

```python
a = "Life is too short, You need python"
print(a[0])
print(a[-2])
```

```python
# 문자열 슬라이싱
print(a[0:4])
print(a[:4])
print(a[19:])
print(a[19:-7])
```

### 1.2.3. 문자열 다루기

```python
# 원하는 문자를 입력하고 실행
text = "Hello, Python!"

print("글자 수:", len(text))
print("대문자로 변환:", text.upper())
print("소문자로 변환:", text.lower())
print("글자 바꾸기:", text.replace("!", "?"))
print("'P'의 위치:", text.find('o'))
print("'z'의 위치:", text.find('z'))
print("공백 제거:", text.replace(" ", ""))
```

```python
print("2026-03-16"[5:7])
print("파이썬 수업자료.pdf"[-3:])
print("260316-3123412"[7])
```

```python
# 사용자가 작성한 아이디와 비밀번호
input_id = " PythonMaster "
input_pw = "password123"

# 불필요한 공백제거
cleaned_id = input_id.replace(" ", "")
print("공백 제거:", cleaned_id)

# 아이디 길이 제한
# 아이디가 5글자 이상, 15글자 이하인지 확인
print("아이디 길이:", len(cleaned_id))

# 아이디를 소문자로 통일
# Python과 python을 같은 아이디로 취급하기 위해 모두 소문자로 변환
final_id = cleaned_id.lower()
print("최종 아이디:", final_id)

# 비밀번호에 특수문자 "!" 포함 여부 확인
print("'!' 포함 여부 (위치):", input_pw.find('!'))
```

```python
### 꿀팁!! f-string(포매팅)
# 변수와 문자열을 섞어서 출력할 때, 쉼표(,)를 사용하는 것보다 f-string을 사용하는 것이 효율적

user_name = "홍길동"
my_age    = 19

# 기본 print 출력
print("제 이름은", user_name, "이고, 나이는", my_age, "살 입니다.")
# "+"를 사용해서 출력
print("제 이름은 " + user_name + "이고, 나이는 " + str(my_age) + "살 입니다.")
# f-string을 사용한 출력
print(f"제 이름은 {user_name}이고, 나이는 {my_age}살 입니다.")
```
